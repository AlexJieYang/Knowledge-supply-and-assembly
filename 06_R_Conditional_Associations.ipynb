# 06 Conditional association models in R

Use the R project kernel after setup_R.R. Paper models include field-year and publication-source fixed effects. Aggregate models use assembler-exposure weights. The original optional AI-profile specification remains available.

```r
candidates <- c(Sys.getenv('GKA_PROJECT', unset = getwd()), getwd(), dirname(getwd()))
valid <- vapply(candidates, function(p) file.exists(file.path(p, 'R', 'analysis.R')), logical(1))
if (!any(valid)) stop('Set the working directory to the repository root.')
PROJECT <- normalizePath(candidates[which(valid)[1]], winslash = '/', mustWork = TRUE)
source(file.path(PROJECT, 'R', 'analysis.R'), encoding = 'UTF-8')
models <- run_associations(PROJECT)
```
