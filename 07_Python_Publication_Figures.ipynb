# 07 Main publication figures

Generate five main figures from completed results: role maps, aligned networks, distributions, role profiles and temporal comparisons. Export PDF, SVG and 450-dpi PNG with captions and panel-to-source mappings.

```python
# Select the project Python kernel before running this cell.
from pathlib import Path
import os, sys, json
candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT = next((p for p in candidates if (p / 'assembly' / 'runtime.py').is_file()), None)
if PROJECT is None:
    raise FileNotFoundError('Open this notebook from the repository root or a subdirectory.')
if str(PROJECT) not in sys.path:
    sys.path.insert(0, str(PROJECT))
CONFIG_PATH = Path(os.environ.get('GKA_CONFIG', str(PROJECT / 'config.json'))).expanduser().resolve()
settings = json.loads(CONFIG_PATH.read_text(encoding='utf-8-sig'))
os.environ['ASSEMBLY_THREADS'] = str(settings['threads'])
import bootstrap
bootstrap.configure(settings['threads'])
from assembly.runtime import config, run, dirs
C = config(CONFIG_PATH)
D = dirs(C)
print('Project:', PROJECT)
print('Python:', sys.executable)
print('Output:', D['root'])
```

```python
run(C, 'figures')
```

```python
from IPython.display import Image, display
for image in sorted(D['figures'].glob('Fig*.png')):
    print(image.name)
    display(Image(filename=str(image), width=1000))
```
