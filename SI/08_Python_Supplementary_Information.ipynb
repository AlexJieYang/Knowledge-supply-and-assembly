# 08 Supplementary analyses

Build supplementary tables, figures and source data from existing main results. Read SI_README.md for workflow numbering and optional inputs.

```python
from pathlib import Path
import os, sys
for name in ('OPENBLAS_NUM_THREADS', 'MKL_NUM_THREADS', 'OMP_NUM_THREADS'):
    os.environ[name] = '1'
candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT = next((p for p in candidates if (p / 'si_support' / 'core.py').is_file()), None)
if PROJECT is None:
    raise FileNotFoundError('Open this notebook from the repository root or its SI directory.')
if str(PROJECT) not in sys.path:
    sys.path.insert(0, str(PROJECT))
from si_support.core import Context, MissingInput
CONFIG_PATH = Path(os.environ.get('GKA_CONFIG', str(PROJECT / 'config.json'))).resolve()
CTX = Context(config=CONFIG_PATH)
print('Input:', CTX.results)
print('SI output:', CTX.out)
```

Reuse summaries, network caches and the plotting sample from stages 01-05. Existing R results are imported when available.

```python
from si_support.analysis import build_tables
build_tables(CTX)
```

Disabled by default. Stream assembly_papers.parquet; source-type diagnostics require aligned kind/ID caches.

```python
RUN_DEEP_AUDIT = False
if RUN_DEEP_AUDIT:
    from si_support.analysis import deep_audit
    try:
        deep_audit(CTX, batch_rows=250000)
    except MissingInput as exc:
        CTX.event('tables', 'deep_audit', 'skipped', exc)
    else:
        CTX.event('tables', 'deep_audit', 'complete')
```
